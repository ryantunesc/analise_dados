```python
import pandas as pd;
import openpyxl ;
path1 = r"H:\python\3etapa\git\analise_dados\atividade_6\Tabela5-sem_emprego_2012.csv"  
path2 = r"H:\python\3etapa\git\analise_dados\atividade_6\Tabela5-sem_emprego_2026.csv"
path3 = r"H:\python\3etapa\git\analise_dados\atividade_6\Tabela 1.1.1.xls"
t2012 = pd.read_csv(path1,encoding = 'utf-8', sep = ";", decimal = ",", skiprows=0)
t2026 = pd.read_csv(path2,encoding = 'utf-8', sep = ";", decimal = ",", skiprows=0)
tabela2 = pd.read_excel(path3, engine="xlrd")
```

```python
tabela2 = tabela2.iloc[8:40].copy()
tabela2.columns = [
    "Estado",
    "total",
    "total_branca",
    "total_preta_parda",
    "homem_branca",
    "homem_preta_parda",
    "mulher_branca",
    "mulher_preta_parda",
]
tabela2 = tabela2.reset_index(drop=True)

# horas são número, não texto
for col in tabela2.columns[1:]:
    tabela2[col] = pd.to_numeric(tabela2[col], errors="coerce")

tabela2
```

```python
comp = t2012.merge(t2026, on=["Sigla", "Código", "Estado"], how="inner")
```

```python
comp2 = comp.merge(tabela2, on=["Estado"], how="inner")
```
