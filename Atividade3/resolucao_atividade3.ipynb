```python
import pandas as pd

print("pandas:", pd.__version__)
```

```python
arquivo = "Tabela_3_Areas_gerais_de_formacao_na_graduacao (1).xlsx"

bruto = pd.read_excel(arquivo, header=None)
bruto.head(10)
```

```python
colunas = [
    "Local",
    "Total_Geral", "Homens_Geral", "Mulheres_Geral",
    "Total_CTEM", "Homens_CTEM", "Mulheres_CTEM",
    "Total_Educ", "Homens_Educ", "Mulheres_Educ",
]


dados = bruto.iloc[3:36].copy()
dados.columns = colunas
dados = dados.reset_index(drop=True)
dados
```

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]

df_regioes = dados[dados["Local"].isin(regioes)].reset_index(drop=True)
df_regioes = df_regioes.rename(columns={"Local": "Região"})


colunas_numericas = df_regioes.columns.drop("Região")
df_regioes[colunas_numericas] = df_regioes[colunas_numericas].astype(int)

df_regioes
```

```python
tabela_superior = df_regioes[["Região", "Homens_Geral", "Mulheres_Geral"]].rename(
    columns={"Homens_Geral": "Homens", "Mulheres_Geral": "Mulheres"}
)
tabela_superior
```

```python
tabela_ctem = df_regioes[["Região", "Homens_CTEM", "Mulheres_CTEM"]].rename(
    columns={"Homens_CTEM": "Homens", "Mulheres_CTEM": "Mulheres"}
)
tabela_ctem
```

```python
tabela_educ_saude = df_regioes[["Região", "Homens_Educ", "Mulheres_Educ"]].rename(
    columns={"Homens_Educ": "Homens", "Mulheres_Educ": "Mulheres"}
)
tabela_educ_saude
```

```python
tabela_superior.to_excel("tabela_curso_superior.xlsx", index=False)
tabela_ctem.to_excel("tabela_ciencia_tecnologia_engenharia_matematica.xlsx", index=False)
tabela_educ_saude.to_excel("tabela_educacao_servicos_saude_bemestar.xlsx", index=False)

print("Arquivos salvos:")
print("- tabela_curso_superior.xlsx")
print("- tabela_ciencia_tecnologia_engenharia_matematica.xlsx")
print("- tabela_educacao_servicos_saude_bemestar.xlsx")
```
